```python
# polynomial
import matplotlib.pyplot as plt
x = [1,2,3,5,6,7,8,9,10,12,13,14,15,16,18,19,21,22]
y = [100,90,80,60,60,55,60,65,70,70,75,76,78,79,90,99,99,100]
plt.scatter(x, y)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# --- REPLACE these with your real data ---
# time in hours since midnight (e.g. 8.5 = 8:30 AM), speed in km/h
X = np.array([6, 8, 10, 12, 14, 16]).reshape(-1, 1)
y = np.array([80, 55, 70, 75, 65, 50])

model = LinearRegression()
model.fit(X, y)

# --- 5:00 PM = 17.0 hours ---
t_pred = np.array([[17.0]])
speed_pred = model.predict(t_pred)[0]

# 1. Print the predicted speed
print(f"Predicted speed at 17:00: {speed_pred:.2f}")

# 2. Display the predicted value on the graph
plt.figure(figsize=(10, 6))
plt.scatter(X, y, label="Observed data")
xs = np.linspace(X.min(), max(X.max(), 17), 200).reshape(-1, 1)
plt.plot(xs, model.predict(xs), label="Model fit")
plt.scatter(17.0, speed_pred, color="red", s=100, zorder=5,
            label=f"Prediction at 17:00 ({speed_pred:.1f})")
plt.annotate(f"{speed_pred:.1f}", (17.0, speed_pred),
             textcoords="offset points", xytext=(10, 10))
plt.xlabel("Time (hours)")
plt.ylabel("Speed")
plt.legend()
plt.show()
```
